### 2.Compute:

### Norm-1 & Norm-2 of vectors (Three students)

```python
Alice = np.array(Data.loc[Data.Name == "Alice",subjects].values[0], dtype=float)
Bob = np.array(Data.loc[Data.Name == "Bob",subjects].values[0], dtype=float)
Charlie = np.array(Data.loc[Data.Name == "Charlie", subjects].values[0], dtype=float)

Alice_L1 = np.linalg.norm(Alice,ord=1);  
Alice_L2 = np.linalg.norm(Alice,ord=2)
Bob_L1 = np.linalg.norm(Bob,ord=1);  
Bob_L2 = np.linalg.norm(Bob,ord=2)
Charlie_L1 = np.linalg.norm(Charlie,ord=1);  
Charlie_L2 = np.linalg.norm(Charlie,ord=2)

print(f"{'Student':<10} {'Vector':<40} {'Norm-1':>8} {'Norm-2':>8}")
print()
print(f"{'Alice':<10} {str(Alice):<40} {Alice_L1:>8.2f} {Alice_L2:>8.2f}")
print(f"{'Bob':<10} {str(Bob):<40} {Bob_L1:>8.2f} {Bob_L2:>8.2f}")
print(f"{'Charlie':<10} {str(Charlie):<40} {Charlie_L1:>8.2f} {Charlie_L2:>8.2f}")
```

- Norm-1 is the total score sum; Norm-2 is the Euclidean length of the score vector.
- A higher Norm-2 = overall stronger student performance.

### Dot product and Angle between two student's score vectors

```python
Alice   = np.array(Data.loc[Data.Name == "Alice", subjects].values[0], dtype=float)
Bob = np.array(Data.loc[Data.Name == "Bob", subjects].values[0], dtype=float)

dot_product = np.dot(Alice, Bob)
dot_product
```

- A small angle (close to 0°) means Alice and Bob have similar score distributions across subjects.

```python
Alice = np.array(Data.loc[Data.Name == "Alice", subjects].values[0], dtype=float)
Bob = np.array(Data.loc[Data.Name == "Bob", subjects].values[0], dtype=float)

magnitude_A = np.linalg.norm(Alice)
magnitude_B = np.linalg.norm(Bob)

cos_theta = dot_product / (magnitude_A * magnitude_B)
angle_radian = np.arccos(cos_theta)  

print(angle_radian)

angle_degree = np.degrees(angle_radian)  

print(angle_degree)
```

- A large angle means very different performance patterns.

### Cross Product(For 3D selected subjects).

```python
Alice_3D = Alice[:3]   
Bob_3D = Bob[:3] 

print(f"Alice (3D) : {Alice_3D} → [Math, Physics, Chemistry]")
print(f"Bob (3D) : {Bob_3D} → [Math, Physics, Chemistry]")
print()

cross_numpy = np.cross(Alice_3D, Bob_3D)
print(f"Cross Product (Alice × Bob) : {cross_numpy}")
print(f"i component : {cross_numpy[0]:.2f}")
print(f"j component : {cross_numpy[1]:.2f}")
print(f"k component : {cross_numpy[2]:.2f}")
```

- The cross product is perpendicular to both score vectors.
- It's not directly meaningful for scores but confirms the geometric relationship between the two students in 3D subject space.

### 3.Find the projection of of one vector

```python
scalar_projection = dot_product / magnitude_B
vector_projection = (dot_product / (magnitude_B**2)) * Bob

print(f"Scalar Projection of Alice onto Bob : {scalar_projection:.2f}")
print(f"Vector Projection of Alice onto Bob : {vector_projection}")
```

### 5.Explain line,plane and hyperplane with respect to our dataset dimension 

Line → 1D object in 2D space

Plane → 2D surface in 3D space

Hyperplane → n-1 dimensional separator in n-dimensional space

Used in Machine Learning as decision boundary.

### 6.Show how dimensionality increases from 2D to 3D  to higher dimension with hyperplanes.

```python
math_range = np.linspace(40, 100, 100)
physics    = 1.2 * math_range - 20        

plt.figure(figsize=(7, 5))
plt.plot(math_range, physics, color="blue", linewidth=2)
plt.scatter(X[:, 0], X[:, 1], color="red", s=60, zorder=5, label="Students")
plt.title("LINE in 2D  (Physics = 1.2 × Math − 20)")
plt.xlabel("Math Score")
plt.ylabel("Physics Score")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

- The blue line is a linear equation (Physics = 1.2×Math − 20) acting as a separator in 2D space. 
- Red dots are students.
- Points above the line score higher in Physics than the line predicts; points below score lower.

```python
x = np.linspace(40, 100, 30)
y = np.linspace(40, 100, 30)
X, Y = np.meshgrid(x, y)
Z = 0.5*X + 0.5*Y                        

fig = plt.figure(figsize=(5, 5))
ax = fig.add_subplot(111, projection="3d")
ax.plot_surface(X, Y, Z, alpha=0.5, color="cyan")
ax.set_title("Plane in 3D (Chemistry = 0.5×Math + 0.5×Physics)")
ax.set_xlabel("Math")
ax.set_ylabel("Physics")
ax.set_zlabel("Chemistry")
plt.tight_layout()
plt.show()
```

- The cyan surface is a flat plane in 3D space defined by Math, Physics, and Chemistry scores.
- It shows how Chemistry is estimated as an average of the other two — illustrating a 2D decision boundary in 3D.

```python
subjects = ["Math", "Physics", "Chemistry", "Biology", "English"]
scores = Data[subjects].values.astype(float)
total = scores.sum(axis=1)
threshold = total.mean()

x_range = np.linspace(40, 100, 100)
y_line = np.full_like(x_range, threshold)   

plt.figure(figsize=(5, 5))
plt.plot(x_range, y_line, color="red", linewidth=2, label=f"Hyperplane = {threshold:.1f}")
plt.scatter(scores[:, 0], total, color="blue", s=80, zorder=5, label="Students")
plt.title("Hyperplane in 5D (Projected to 2D)")
plt.xlabel("Math Score")
plt.ylabel("Total Score")
plt.legend()
plt.grid(True)
plt.show()
```

- The red horizontal line is the hyperplane (threshold = mean total score).
- Blue dots above it are high-performing students; dots below are low-performers.
- This mimics how ML classifiers draw boundaries in high-dimensional space.

### 8.Perform **LU Decomposition** of the dataset matrix.

```python
from scipy.linalg import lu

P, L, U = lu(X)

reconstructed_A = P @ L @ U

print("Matrix X:\n", X)

print("\nLower Triangular Matrix L:\n", L.round(4))

print("\nUpper Triangular Matrix U:\n", U.round(4))

print("\nReconstructed A (P @ L @ U):\n", reconstructed_A.round(2))
```

- The matrix X is decomposed into Lower (L) and Upper (U) triangular parts.-
- Reconstructed A should match original X exactly — confirming correctness of decomposition.

### 9.Perform **Singular Value Decomposition(SVD)** and explain its role in dimensionality reduction.

```python
U, S, VT = np.linalg.svd(X)

Sigma = np.zeros_like(X, dtype=float)
np.fill_diagonal(Sigma, S)

A_reconstructed = U @ Sigma @ VT

print("Matrix A (Students × Subjects):\n", X)

print("\nMatrix U (10×10):\n", U.round(4))

print("\nSingular Values:", S.round(4))

print("\nMatrix V^T (5×5):\n", VT.round(4))

print("\nReconstructed Matrix:\n", A_reconstructed.round(2))
```

- Singular values (S) represent the importance of each data dimension.
- Large singular values capture most of the data's variance, allowing us to reduce dimensions by keeping only the top components.

## Part E : Dimensionality Reduction

### 10.Apply **Principal Component  Analysis(PCA)** to reduce the dataset from multiple subjects to 2 dimension.

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("Original Data:\n", X)

print("\nStandardized Data:\n", X_scaled.round(4))

print("\nPrincipal Components:\n", pca.components_.round(4))

print("\nExplained Variance Ratio:\n", pca.explained_variance_ratio_.round(4))

print("\nTransformed Data (2D):\n", X_pca.round(4))
```

- PCA compresses 5-subject data into 2 components.
- The explained variance ratio shows how much information is retained.
- If the two components explain ~80%+ variance, the reduction is effective.

### 10.Apply **Linear Discriminant Analysis(LDA)**  to classsify students into "Above Average" and "Below Average" categories.

```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis

X = Data[subjects].values.astype(float)
avg = X.mean(axis=1)
y = np.where(avg >= avg.mean(), 1, 0)
y_labels = np.where(avg >= avg.mean(), "Above Average", "Below Average")
lda = LinearDiscriminantAnalysis()          
X_lda = lda.fit_transform(X, y)

print("Original Data:\n", X)
print("\nClass Labels:\n", y)
print("(0 = Below Average, 1 = Above Average)")
print("\nStudent Categories:")

for name, label in zip(Data["Name"], y_labels):
    print(f"  {name:<10} → {label}")

print("\nLDA Projection:\n", X_lda.round(4))
print("\nLDA Coefficients:\n", lda.coef_.round(4))
```

- LDA projects students onto a single axis that best separates "Above Average" from "Below Average" groups.
- Students with positive LDA scores are above average; negative scores are below.
